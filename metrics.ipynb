```python
start = df.groupby(['Ticker', 'year']).first().reset_index()[['Ticker', 'year', 'close']]
end = df.groupby(['Ticker', 'year']).last().reset_index()[['Ticker', 'year', 'close']]
# Merge start and end to compute return
returns_df = pd.merge(start, end, on=['Ticker', 'year'], suffixes=('_start', '_end'))
# Calculate % return
returns_df['yearly_return'] = ((returns_df['close_end'] - returns_df['close_start']) / returns_df['close_start']) * 100
latest_year = returns_df['year'].max() #Top 10 Green Stocks (for the most recent year)
top_10_green = returns_df[returns_df['year'] == latest_year].sort_values(by='yearly_return', ascending=False).head(10)
print("✅ Top 10 Green Stocks:")
print(top_10_green[['Ticker', 'year', 'yearly_return']])
```

```python
# Sort and group
df1 = df.sort_values(['Ticker', 'date'])
strt = df1.groupby(['Ticker','year']).first()['close']
end = df1.groupby(['Ticker','year']).last()['close']

# Calculate yearly return
returns = ((end - strt) / strt) * 100
ret_df = returns.reset_index().rename(columns={'close': 'yearly_return'})

# Top 10 Green Stocks
green_10 = ret_df.sort_values(by='yearly_return', ascending=False).head(10)
loss_10 = ret_df.sort_values(by='yearly_return', ascending=False).tail(10)
print("✅ Top 10 Green Stocks:")
print(green_10)
print("✅ Top 10 Loss Stocks:")
print(loss_10)
```

```python
# Sort for group operations
df1 = df1.sort_values(by=['Ticker', 'date'])

# Calculate daily return
df1['daily_return'] = df1.groupby('Ticker')['close'].pct_change()

# Calculate cumulative return (total growth from start to end of year)
df1['avg_yearly_return'] = df1.groupby('Ticker')['daily_return'].transform(lambda x: (1 + x).cumprod() - 1)
```
