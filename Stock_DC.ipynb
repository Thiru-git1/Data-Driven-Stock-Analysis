```python
#save the Ticker data in separate csv files under one folder

import os

# Create a folder to store output files
output_folder = 'nifty50'
os.makedirs(output_folder, exist_ok=True)

# Loop through each unique ticker
for ticker in df['Ticker'].unique(): # here picks each unique ticker name is SBIN', 'BAJFINANCE', 'TITAN', 'ITC ...
    ticker_df = df[df['Ticker'] == ticker]  # filter dat
    file_name = f"{ticker}.csv"             # create filename like SBIN.csv...so on...
    file_path = os.path.join(output_folder, file_name)  # full path
    ticker_df.to_csv(file_path, index=False)  # save CSV
    print(f"Saved: {file_name}")
```

```python
ticker_map = {ticker: i for i, ticker in enumerate(df['Ticker'].unique(), start=1)}
df['stock_no'] = df['Ticker'].map(ticker_map)

# Reorder columns to make 'stock_no' the first column
cols = ['stock_no'] + [col for col in df.columns if col != 'stock_no']
df = df[cols]
```

```python
df['month'] = df['date'].dt.month
```

```python
df['month'] = df['month'].astype(int)
```

```python
df.to_csv("nifty50_cleaned.csv", index=False)
```
